```python
# ==============================
# 1. IMPORT LIBRARIES
# ==============================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# ==============================
# 2. DEFINE PATHS
# ==============================
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_PATH = PROJECT_ROOT / "data" / "train_FD001.txt"
CSV_DATA_PATH = PROJECT_ROOT / "data" / "train_FD001.csv"

print("Raw TXT path  :", RAW_DATA_PATH)
print("CSV output path:", CSV_DATA_PATH)
```

```python
# ==============================
# 3. LOAD DATA
# ==============================
df = pd.read_csv(
    RAW_DATA_PATH,
    sep=r"\s+",
    header=None
)

print("✅ Raw data loaded successfully")
df.head()
```

```python
# ==============================
# 4. ASSIGN COLUMN NAMES
# ==============================
columns = (
    ["engine_id", "cycle"] +
    [f"op_setting_{i}" for i in range(1, 4)] +
    [f"sensor_{i}" for i in range(1, 22)]
)

df.columns = columns

print("✅ Columns assigned successfully")
df.head()
```

```python
# ==============================
# 5. SAVE AS CSV (CLEAN FORMAT)
# ==============================
df.to_csv(CSV_DATA_PATH, index=False)

print("✅ CSV file saved at:", CSV_DATA_PATH)
```

```python
# ==============================
# 6. DOMAIN SANITY CHECKS  
# ==============================
print("Unique engines  :", df["engine_id"].nunique())
print("Cycle min per engine (expect 1):")
print(df.groupby("engine_id")["cycle"].min().unique())  # should be [1]
print("\nEngine lifecycle summary:")
print(df.groupby("engine_id")["cycle"].max().describe())
```

```python
# ==============================
# 7. CREATE TARGET VARIABLE (RUL)
# ==============================

# Step 1: Get max cycle for each engine
max_cycle = df.groupby("engine_id")["cycle"].max().reset_index()
max_cycle.columns = ["engine_id", "max_cycle"]

# Step 2: Merge with original data
df = df.merge(max_cycle, on="engine_id")

# Step 3: Calculate RUL
df["RUL"] = df["max_cycle"] - df["cycle"]

# Step 4: Drop unnecessary column
df.drop("max_cycle", axis=1, inplace=True)

print("✅ RUL column created successfully")
df.head()
```

```python
# ==============================
# 9. DOMAIN UNDERSTANDING
# ==============================
"""
Dataset: NASA CMAPSS FD001 — Turbofan Engine Degradation
- engine_id    : Unique engine identifier
- cycle        : Operational time step
- op_setting_1-3 : Operational conditions (FD001 = single condition)
- sensor_1-21  : Raw sensor measurements
- RUL          : Remaining Useful Life (target) — cycles left until failure

All 100 engines are run to failure (RUL = 0 at last cycle).
Task: Regression — predict RUL from sensor readings.
"""
```

```python
# ==============================
# 10. OUTLIER DETECTION
# ==============================
feature_cols_for_outlier = (
    [f"op_setting_{i}" for i in range(1, 4)] +
    [f"sensor_{i}" for i in range(1, 22)]
)

Q1  = df[feature_cols_for_outlier].quantile(0.25)
Q3  = df[feature_cols_for_outlier].quantile(0.75)
IQR = Q3 - Q1

outliers = (
    (df[feature_cols_for_outlier] < (Q1 - 1.5 * IQR)) |
    (df[feature_cols_for_outlier] > (Q3 + 1.5 * IQR))
).sum()

print("Outlier count per column (sensor/op features only):\n")
print(outliers.sort_values(ascending=False))


# NOTE: High "outlier" counts in sensor_9 / sensor_14 are likely real
# degradation signals, NOT noise — do not blindly remove them.
```

```python
# ==============================
# 11. SAVE UPDATED DATA
# ==============================
df.to_csv(CSV_DATA_PATH, index=False)
print("✅ Final dataset with RUL saved")
```
